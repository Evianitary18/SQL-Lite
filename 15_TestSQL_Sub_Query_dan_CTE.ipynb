Keuntungan menggunakan Subquery
1. memungkinkan kita untuk mengisolasi setiap bagian dari suatu statement.
2. sebagai alternatif yang lebih simpel untuk query data dari tabel, dibandingkan JOIN dan UNION yang lebih kompleks.
3. Memungkinkan kita untuk menyalin data dari suatu tabel ke tabel lain.
4. Memungkinkan kita untuk mengambil data dari table lain untuk ditambahkan, diperbarui, atau dihapus pada tabel tujuan.

Aturan menggunakan SUBQUERY:
1. Subquery harus berada di dalam parentheses (tanda kurung).
2. Dapat menggunakan operator pembanding seperti ">", "<", "=", IN, ANY, SOME, dll.

Letak SUBQUERY
1. SELECT statement
digunakan ketika ingin melakukan perhitungan dengan menggunakan fungsi agregat, tetapi tidak ingin menerapkan fungsi agregat ke query utama.

SELECT nama_kolom,
(SELECT ... FROM ...) AS alias
FROM nama_tabel

2. Klausa FROM

SELECT nama_kolom
FROM (SELECT ... FROM ...) AS nama_alias

3. Klausa WHERE

SELECT nama_kolom,
FROM nama_tabel
WHERE (SELECT ...  FROM ...)
